### Age further analysis

```python
age_bins = [10, 20, 30, 40, 50, 60, 70]

# Create age groups using pd.cut
cust_final1['age_group'] = pd.cut(cust_final1['age'], bins=age_bins)

# Group by age group and Y, and count occurrences
age_group_counts = cust_final1.groupby(['age_group', 'Y']).size().unstack(fill_value=0)

# Plot the counts of leaving (1) and stable (0) customers by age group
age_group_counts.plot(kind='bar', width=0.4)
plt.xlabel('Age Group')
plt.ylabel('Count')
plt.title('Distribution of Leaving and Stable Customers by Age Group')
plt.legend(title='Y', labels=['Stable', 'Leaving'])
plt.show()
```
