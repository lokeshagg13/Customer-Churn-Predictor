```python
def extract_tenure(x):
    match = re.findall('(member|customer) for (\d+.\d)', x)
    if match:
        return float(match[0][1])
    else:
        match = re.findall('has been a customer for (\w+|\d+.\d)', x)
        if match:
            try:
                return float(match[0])
            except ValueError:
                if match[0] == 'one':
                    return 1.0
        match = re.findall('has only been with the bank for (\w+|\d+.\d) tenure', x)
        if match:
            try:
                return float(match[0])
            except ValueError:
                if match[0] == 'one':
                    return 1.0
        match = re.findall('has been with the bank for only (\w+|\d+.\d)', x)
        if match:
            try:
                return float(match[0])
            except ValueError:
                if match[0] == 'one':
                    return 1.0
        match = re.findall('has been with the bank for (\d+.\d)', x)
        if match:
            return float(match[0])
        match = re.findall('has been with the company for (\w+|\d+.\d)', x)
        if match:
            try:
                return float(match[0])
            except ValueError:
                if match[0] == 'one':
                    return 1.0
        match = re.findall('has only been with the bank for (\w+)', x)
        if match:
            try:
                return float(match[0])
            except ValueError:
                if match[0] == 'one':
                    return 1.0
    return None
```

```python
df['tenure_from_text'] = df['customer_profile'].apply(extract_tenure)
```

```python
def extract_products(x):
    match = re.findall('(has|holds|has only|only) (\d+\.?\d*|[a-z]+) product[s]?', x)
    if match:
        # If the match is a digit or a decimal number, return it as an integer
        if match[0][1].replace('.', '').isdigit():
            return int(float(match[0][1]))
        # If the match is 'one', return it as an integer
        elif match[0][1] == 'one':
            return 1
    return None
```

```python
df['number_of_products_from_text'] = df['customer_profile'].apply(extract_products)
```

```python
different_rows = df[df['number_of_products_from_text'] != df['number_of_products']]

# Print these rows
len(different_rows)
```

```python
def extract_balance(x):
    match = re.findall('balance of ([\d,]+.\d+)', x)
    if match:
        # Remove the comma and convert to float
        return float(match[0].replace(',', ''))
    else:
        return None

```

```python

df['balance_from_text'] = df['customer_profile'].apply(extract_balance)
```

```python
def extract_salary(x):
    match = re.findall('estimated salary is ([\d,]+.\d+)', x)
    if match:
        # Remove the comma and convert to float
        return float(match[0].replace(',', ''))
    else:
        return None

```
