```python
import pandas as pd
import spacy
from spacy.lang.en.stop_words import STOP_WORDS
```

```python
cust_labelled = pd.read_csv('A2_customer_churn_labeled.csv')
cust_labelled.head()
```

```python
cust_labelled.loc[0, 'customer_profile']
```

```python
# Load spaCy's English tokenizer and tagger
nlp = spacy.load("en_core_web_sm")

# Define a function to perform tokenization, stopwords removal, and lemmatization
def preprocess_text(text):
    doc = nlp(text)
    tokens = [token.lemma_ for token in doc if token.text.lower() not in STOP_WORDS]
    return " ".join(tokens)

preprocess_text(cust_labelled.loc[0, 'customer_profile'])
```

```python
cust_labelled['customer_profile_tokenized'] = cust_labelled['customer_profile'].apply(lambda text: preprocess_text(text))
cust_labelled
```

```python
cust_labelled.to_csv('A2_customer_churn_labeled_NLP.csv', index=False)
```
